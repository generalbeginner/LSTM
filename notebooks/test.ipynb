```python
import sys
import os
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
os.chdir(Path.cwd().parent)

from run_yfinance_lstm import forecast_future
import joblib
import tensorflow as tf
import pandas as pd

ticker = "TSLA"
target = "Close"

model = tf.keras.models.load_model(
    f"models/yfinance_lstm/{ticker.lower()}_{target.lower()}_lstm.keras"
)

scaler = joblib.load(
    f"models/yfinance_lstm/{ticker.lower()}_{target.lower()}_scaler.gz"
)

recent_data = pd.read_csv(
    f"data/yfinance/{ticker.lower()}_raw.csv",
    parse_dates=["Date"]
)

forecast = forecast_future(
    model=model,
    recent_data=recent_data,
    scaler=scaler,
    target=target,
    horizon=20
)

print(forecast)
```
